Employee Attrition Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20)

plt.title('Age Distribution')
plt.show()
```

```python
# Most employees are between 30 and 40 years old.
```

Gender vs Attrition

```python
sns.countplot(
    x='Gender',
    hue='Attrition',
    data=df
)

plt.show()
```

```python
# Compare attrition rates among male and female employees.
```

Department vs Attrition

```python
plt.figure(figsize=(10,5))

sns.countplot(
    x='Department',
    hue='Attrition',
    data=df
)

plt.show()
```

```python
# Identify departments with higher employee turnover.
```

Overtime vs Attrition

```python
sns.countplot(
    x='OverTime',
    hue='Attrition',
    data=df
)

plt.show()
```

```python
# Employees working overtime tend to leave more frequently.
```

Monthly Income vs Attrition

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Attrition',
    y='MonthlyIncome',
    data=df
)

plt.show()
```

```python
# Employees with lower salaries appear more likely to leave.
```

Job Satisfaction vs Attrition

```python
sns.countplot(
    x='JobSatisfaction',
    hue='Attrition',
    data=df
)

plt.show()
```

```python
# Lower job satisfaction is associated with higher attrition.
```

Years at Company vs Attrition

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Attrition', y='YearsAtCompany', data=df)
plt.title('Years at Company vs Attrition')
plt.show()
```

```python
# Employees with fewer years at the company may have higher attrition.
```

Job Role vs Attrition

```python
plt.figure(figsize=(12,6))
sns.countplot(x='JobRole', hue='Attrition', data=df)
plt.xticks(rotation=45)
plt.title('Attrition by Job Role')
plt.show()
```

```python
# Identify roles with the highest turnover.
```

Monthly Income Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['MonthlyIncome'], bins=30)
plt.title('Monthly Income Distribution')
plt.show()
```

Correlation Heatmap

```python
num_df = df.select_dtypes(include='number')

plt.figure(figsize=(15,10))

sns.heatmap(
    num_df.corr(),
    cmap='coolwarm'
)

plt.show()
```

Identify relationships among numerical variables.

# Key Findings

1. Employees working overtime have higher attrition.

2. Lower income employees tend to leave more often.

3. Job satisfaction influences employee retention.

4. Some departments show higher turnover rates.

5. Younger employees are more likely to leave.

# Recommendations

1. Reduce excessive overtime.
2. Improve employee satisfaction programs.
3. Review compensation for lower-income employees.
4. Focus retention efforts on high-attrition departments.
5. Strengthen onboarding and career growth opportunities.

### Conclusion
This analysis identified major factors influencing employee attrition, including overtime, salary, job satisfaction, and tenure. These insights can help organizations improve employee retention strategies.
